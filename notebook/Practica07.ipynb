# 14. Análisis Exploratorio de Datos (EDA)
Distribución de la edad

```python
plt.hist(df["Edad"], bins=20)

plt.title("Distribución de la Edad")

plt.xlabel("Edad")

plt.ylabel("Frecuencia")

plt.show()
```

## Distribución del IMC

```python
plt.hist(df["IMC"], bins=20)

plt.title("Distribución del IMC")

plt.xlabel("IMC")

plt.ylabel("Frecuencia")

plt.show()
```

## Por sexo

```python
df["Sexo"].value_counts().plot(kind="bar")

plt.title("Distribución por Sexo")

plt.xlabel("Sexo")

plt.ylabel("Pacientes")
```

## Distribución por riesgo vascular


```python
import matplotlib.pyplot as plt

df["Tiene_Diabetes"].value_counts().plot(kind="bar")

plt.title("Distribución de pacientes con diabetes")
plt.xlabel("Tiene Diabetes")
plt.ylabel("Cantidad de pacientes")

plt.show()
```

## Distribución de fumadores

```python
df["Fumador"].value_counts().plot(kind="bar")

plt.title("Pacientes Fumadores")

plt.show()
```

## Actividad fisica

```python
import matplotlib.pyplot as plt

df["Actividad_Fisica"].value_counts().plot(kind="bar")

plt.title("Nivel de Actividad Física de los Pacientes")
plt.xlabel("Actividad Física")
plt.ylabel("Cantidad de pacientes")

plt.show()
```

## Glucosa ayuno

```python
import matplotlib.pyplot as plt

plt.scatter(
    df["Edad"],
    df["Glucosa_Ayuno"]
)

plt.title("Edad vs Glucosa en Ayuno")

plt.xlabel("Edad")
plt.ylabel("Glucosa en ayuno (mg/dL)")

plt.show()
```

## edad vs glucosa

```python
import matplotlib.pyplot as plt

plt.scatter(
    df["IMC"],
    df["Glucosa_Ayuno"]
)

plt.title("IMC vs Glucosa en Ayuno")

plt.xlabel("IMC")
plt.ylabel("Glucosa en ayuno (mg/dL)")

plt.show()
```
